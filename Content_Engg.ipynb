```python
from google.colab import userdata

GROQ_API_KEY = userdata.get("GROQ_API_KEY")
print("API keys loaded successfully!")
```

```python
from groq import Groq

client = Groq(api_key=GROQ_API_KEY)
MODEL = "openai/gpt-oss-120b"
print("NutriAI model initialized successfully!")
```

```python
# This list keeps track of your conversation context
if 'chat_history' not in locals() and 'chat_history' not in globals():
    chat_history = []
    print("Conversation memory initialized!")
```

```python
def understand_meal_with_memory(user_input, history_list):
    system_prompt = """
    You are a professional Nutrient AI assistant.

    Rules:
    1. Act as a professional and always remember the user's details (such as their name) if they provide it.
    2. Identify food names and quantities from meal descriptions.
    3. Convert serving quantities into estimated grams.
    4. For one medium bowl of cooked rice, assume 150 grams.
    5. For one omelette without egg count, assume 2 eggs.
    6. For one omelette, estimate 5 grams cooking oil.
    7. Break mixed dishes into ingredients where appropriate.
    8. If the user says only '1 bowl' without identifying what is inside it, ask for clarification.
    9. Cooked white rice calories should be calculated based on 100 kcal per 100g.
    """

    messages = [{"role": "system", "content": system_prompt}]
    messages.extend(history_list)
    messages.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        model=MODEL,
        messages=messages,
        temperature=0.2
    )

    assistant_reply = response.choices[0].message.content

    # Update chat history
    history_list.append({"role": "user", "content": user_input})
    history_list.append({"role": "assistant", "content": assistant_reply})

    return assistant_reply

def analyze_meal_v2(user_input):
    global chat_history
    analysis = understand_meal_with_memory(user_input, chat_history)
    print(analysis)
    return analysis
```

```python
analyze_meal_v2("Hi, my name is Naveen.")
```
